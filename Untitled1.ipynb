```python
# to find the second largest unique element

numbers = [12, 45, 23, 45, 67, 12]

largest = float('-inf')
second_largest = float('-inf')

for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif largest > num > second_largest:
        second_largest = num

if second_largest == float('-inf'):
    print("Second largest unique element does not exist.")
else:
    print("Second largest unique element:", second_largest)
```

```python
#rotate a list to the right by k positions

numbers = [10, 20, 30, 40, 50]
k = 2
k = k % len(numbers)

rotated = numbers[-k:] + numbers[:-k]

print("Rotated List:", rotated)
```

```python
#  to remove duplicates while preserving order

numbers = [10, 20, 30, 20, 40, 10, 50, 30]

unique = []
seen = set()

for num in numbers:
    if num not in seen:
        unique.append(num)
        seen.add(num)

print("List after removing duplicates:", unique)
```

```python
#the length of the longest consecutive sequence.
nums = [100, 4, 200, 1, 3, 2]

num_set = set(nums)
longest = 0

for num in num_set:

    # Check if this is the start of a sequence
    if num - 1 not in num_set:

        current = num
        length = 1

        while current + 1 in num_set:
            current += 1
            length += 1

        longest = max(longest, length)

print("Length of longest consecutive sequence:", longest)
```

```python
# Read matrix size
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))
48
matrix1 = []
print("Enter first matrix:")
for i in range(rows):
    row = []
    for j in range(cols):
        row.append(int(input()))
    matrix1.append(row)

matrix2 = []
print("Enter second matrix:")
for i in range(rows):
    row = []
    for j in range(cols):
        row.append(int(input()))
    matrix2.append(row)

# Matrix Addition
addition = []
for i in range(rows):
    row = []
    for j in range(cols):
        row.append(matrix1[i][j] + matrix2[i][j])
    addition.append(row)

# Matrix Subtraction
subtraction = []
for i in range(rows):
    row = []
    for j in range(cols):
        row.append(matrix1[i][j] - matrix2[i][j])
    subtraction.append(row)

# Matrix Multiplication
multiplication = []

for i in range(rows):
    row = []
    for j in range(cols):
        total = 0
        for k in range(cols):
            total += matrix1[i][k] * matrix2[k][j]
        row.append(total)
    multiplication.append(row)

# Display results
print("\nAddition:")
for row in addition:
    print(row)

print("\nSubtraction:")
for row in subtraction:
    print(row)

print("\nMultiplication:")
for row in multiplication:
    print(row)
```

```python
# Read the list
numbers = list(map(int, input("Enter elements: ").split()))

n = len(numbers)
majority = None

for num in numbers:
    count = numbers.count(num)

    if count > n // 2:
        majority = num
        break

if majority is not None:
    print("Majority Element:", majority)
else:
    print("No Majority Element")
```
