```python
#frequencyof each element in a list
numbers=[12,45,23,45,67,12]
frequency={}
for num in numbers:
  if num in frequency:
    frequency[num] += 1
  else:
    frequency[num] = 1
print("frequency of each element:")
for key in frequency:
  print(key,":",frequency[key])
```

```python
#found target value at index
integers = [10,20,50,60,40]
target_value = 20
if target_value in integers:
    print("target value",target_value,"is at index",integers.index(target_value))
else:
    print("value of target not found")

```

```python
#list of integers and a target value and whose sum equals to target
integer=[12,24,36,48]
target = 60
for i in range(len(integer)):
  for j in range(i+1,len(integer)):
      if integer[i]+integer[j]==target:
          print(integer[i],integer[j])
```

```python
#second largest unique number
num=[12,45,23,45,67,12]
num1=[]
for x in range(len(num)):
    count=0
for a in num:
    if num[x]==a:
        count+=1
    if count==1:
        num1.append(num[x])
print(num1)
for x in range(len(num1)-1):
    if num1[x]<num1[x+1]:
        num1[x],num1[x+1]=num1[x+1],num1[x]
print(num1[1])

```
